# Defining compounds for property prediction and molecular design

It is important that compinds defined are validated. this validation include:
1. making sur the groups used are indeed unifac groups
2. making sure the defined compound adheres to the structural constraints as defined by [Odele and Macchietto](https://www.sciencedirect.com/science/article/abs/pii/037838129387127M)
In the following, we will define a series of compound and try to validate these.

## Examples: invalid compound
we will define compund that uses a group called "CH5", which does not exists

```python
# imports to enable the notebook to access the src folder
import sys
import os
sys.path.append(os.path.dirname(os.getcwd()))

# import the compound validator
from src.molecules import StructuralValidator
validator = StructuralValidator()


# define compound as a dict:
compound_wrong_group = {'CH5':2}

# test for validity
validator.is_valid(compound_wrong_group)
```

As can be seen the code returns false. Now lets try a compound where we do not respect the constraints of Odele and Macchietto.

```python
# define 2 compounds that does fulffill the structural constraints
compounds = {'invalid_1': {'CH3':1, 'CH2':1},
             'invalid_2': {'CH3':1, 'CH2': 3, 'H2O':1, 'CHO': 1}}
# test for valdidity
validator.validate_batch(compounds)
```

compound 1, has unused valency while the second compound contains water in it, which is a complete compound on its own.


## Examples: valid compound
we define a few valid compounds to test the structural validation.

```python
# define compounds
compounds = {'ethanol': {'CH3':1, 'CH2':1, 'OH':1},
             'methanol': {'CH3OH':1},
             'acetone': {'CH3':1, 'CH3CO':1},
             'heptane': {'CH3':2, 'CH2':5},
             'water': {'H2O':1},
             'DEA': {'CH3':2, 'CH2':1, 'CH2NH':1}}
# test the validity
validator.validate_batch(compounds)
```
