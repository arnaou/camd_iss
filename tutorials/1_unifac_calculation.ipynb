# GC calculation using Constantinou-Gani method

## Introduction
This notebook aims to introduce the group-contribution methods developed by, based on or compatible with [Constantinou and Gani (1994)](https://aiche.onlinelibrary.wiley.com/doi/abs/10.1002/aic.690401011). Some of the related work also covers [Contantinou et al. (1995)](https://www.sciencedirect.com/science/article/pii/037838129402593P) and [Martin and Young (2001)](https://pubs.acs.org/doi/10.1021/tx0155045).
