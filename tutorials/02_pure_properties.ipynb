# Calculation of pure component properties
In this demo, we will use the Constantinou-Gani group-contribution method.

Let's load the necessairy modules.

```python
# imports to enable the notebook to access the src folder
import sys
import os
sys.path.append(os.path.dirname(os.getcwd()))

# import validator (important) and the property calculator
from src.pure import PropertyCalculator

# Using default method (Constantinou-Gani)
calc = PropertyCalculator()
```

Let us first check what properties are available and their units.

```python
print("Available properties:", calc.method.get_available_properties())
print("Property units:", calc.method.get_property_units())
print("Method name:", calc.method.get_method_name())
```

We will test the tools for two cases:
* single compound calculation
* batch calculation

# Single compound
Lets us test the tool on a single compound asking for all properties.

```python
compound = {'CH3': 1, 'CH2': 1, 'OH': 1}  # Ethanol
properties = calc.calculate(compound)  # Auto-detects single compound
properties
```

We can also just ask for a a selected few properties

```python
selected_properties = calc.calculate(compound, properties=['tc', 'pc', 'tb', 'ac'])  # Auto-detects single compound
selected_properties
```

```python
compounds = {
    'water': {'H2O': 1},
    'ethanol': {'CH3': 1, 'CH2': 1, 'OH': 1},
    'acetone': {'CH3': 1, 'CH3CO': 1},
    'methanol': {'CH3OH': 1}
}
batch_results = calc.calculate(compounds)  # Auto-detects batch
batch_results
```

we know similarely only as for a selected set of properties
