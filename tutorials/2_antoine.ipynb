# Antoine Equation Parameter Estimation

## Introduction

This notebook introduces the procedure for estimation the Antoine parameters using pure component properties. The Antoine equation is used to calculate the saturated vapour pressure. The equation for the Antoine equation can be seen below:
$$
\log_{10}(P) = A - \dfrac{B}{T+C}
$$

Consider a phase diagram as seen below, the various properties correspond to points situated on the vapor saturation line.

![phase diagram](../img/phase_diagram.png)

As a consequence, The Antoine parameters can be determined using the following properties: the normal boiling point (Tb), the critical temperature and pressure (Tc and Pc) as well as the acentric factor. The Acentric factor is calculated as shown below:
$$
\omega = - \log_{10}(\left P^*_r \right)-1
$$
with
$$
P^*_r = \dfrac{P^*}{P_c} \qquad T_r = \dfrac{T}{T_c} = 0.7
$$

Accordingly, we then have information about three points situated on the vapour saturation line:

| no. 	| T       	| P    	  |
|-----	|---------	|---------|
| 1   	| Tb      	| Patm 	  |
| 2   	| T=0.7Tc 	| P$^*$  	|
| 3   	| Tc      	| Pc   	  |

## Procedure
The determination of the Antoine parameters consists of solving the following system of equations
\begin{align}
\log_{10}(P_{atm}) &= A - \dfrac{B}{T_b+C} \\
\log_{10}(P_c) &= A - \dfrac{B}{T_c+C}\\
\log_{10}(P^*) &= A - \dfrac{B}{0.7Tc+C}

\end{align}
